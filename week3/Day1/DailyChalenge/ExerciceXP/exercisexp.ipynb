```python
import numpy as np

# Exercise 1: Array Creation and Manipulation
print("Exercise 1:")
arr1 = np.arange(0, 10)
print(arr1)
print()

# Exercise 2: Type Conversion and Array Operations
print("Exercise 2:")
arr2 = np.array([3.14, 2.17, 0, 1, 2]).astype(int)
print(arr2)
print()

# Exercise 3: Working with Multi-Dimensional Arrays
print("Exercise 3:")
arr3 = np.arange(1, 10).reshape(3, 3)
print(arr3)
print()

# Exercise 4: Creating Multi-Dimensional Array with Random Numbers
print("Exercise 4:")
arr4 = np.random.rand(4, 5)
print(arr4)
print()

# Exercise 5: Indexing Arrays
print("Exercise 5:")
arr5 = np.array([[21, 22, 23, 22, 22], [20, 21, 22, 23, 24], [21, 22, 23, 22, 22]])
second_row = arr5[1]
print(second_row)
print()

# Exercise 6: Reversing elements
print("Exercise 6:")
arr6 = np.arange(0, 10)
arr6_reversed = arr6[::-1]
print(arr6_reversed)
print()

# Exercise 7: Identity Matrix
print("Exercise 7:")
arr7 = np.eye(4)
print(arr7)
print()

# Exercise 8: Simple Aggregate Funcs
print("Exercise 8:")
arr8 = np.arange(0, 10)
sum_val = np.sum(arr8)
avg_val = np.average(arr8)
print(f"Sum: {sum_val}, Average: {avg_val}")
print()

# Exercise 9: Create Array and Change its Structure
print("Exercise 9:")
arr9 = np.arange(1, 21).reshape(4, 5)
print(arr9)
print()

# Exercise 10: Conditional Selection of Values
print("Exercise 10:")
arr10 = np.arange(0, 10)
odd_numbers = arr10[arr10 % 2 != 0]
print(odd_numbers)
```
