```python
import numpy as np
import pandas as pd

import matplotlib.pyplot as plt

np.random.seed(0)

cities = [f"City {i}" for i in range(1, 11)]
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun", "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]

temperature_data = np.random.uniform(-5, 35, size=(len(cities), len(months)))
df = pd.DataFrame(temperature_data, index=cities, columns=months)

annual_avg = df.mean(axis=1)
city_highest = annual_avg.idxmax()
city_lowest = annual_avg.idxmin()
highest_avg = annual_avg.max()
lowest_avg = annual_avg.min()

print("Annual average temperature by city:")
print(annual_avg.round(2))
print()
print(f"Highest average temperature: {city_highest} ({highest_avg:.2f} °C)")
print(f"Lowest average temperature: {city_lowest} ({lowest_avg:.2f} °C)")

plt.figure(figsize=(12, 6))
for city in cities:
    plt.plot(months, df.loc[city], marker="o", label=city)
plt.title("Monthly Average Temperatures for 10 Cities")
plt.xlabel("Month")
plt.ylabel("Temperature (°C)")
plt.grid(alpha=0.4)
plt.legend(bbox_to_anchor=(1.05, 1), loc="upper left")
plt.tight_layout()
plt.show()

plt.figure(figsize=(8, 5))
annual_avg.sort_values().plot(kind="bar", color="skyblue")
plt.title("Annual Average Temperature by City")
plt.ylabel("Temperature (°C)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
