```python
# Exercise 2: Descriptive Statistics
# Task: Given a sample dataset, calculate the mean, median, variance, and standard deviation using SciPy.
# Sample Dataset:
import numpy as np

data = [12, 15, 13, 12, 18, 20, 22, 21]
# Calculate Mean
mean = np.mean(data)
# Calculate Median
median = np.median(data)
# Calculate Variance
variance = np.var(data, ddof=1)  # Sample variance
# Calculate Standard Deviation
std_dev = np.std(data, ddof=1)  # Sample standard deviation
print("Mean:", mean)
print("Median:", median)
print("Variance:", variance)
print("Standard Deviation:", std_dev)
```

```python
# Exercise 3: Understanding Distributions
# Task: Generate a normal distribution using SciPy with a mean of 50 and a standard deviation of 10.
#  Plot this distribution.
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm
# Parameters for the normal distribution
mean = 50
std_dev = 10
# Generate a range of x values
x = np.linspace(mean - 4*std_dev, mean + 4*std_dev, 1000)
# Generate the normal distribution values
y = norm.pdf(x, loc=mean, scale=std_dev)
# Plot the distribution
plt.plot(x, y)
plt.xlabel("Value")
plt.ylabel("Probability Density")
plt.title("Normal Distribution")
plt.show()
```

```python
# Exercise 4: T-Test Application
# Perform a T-test on two sets of randomly generated data.
from scipy import stats
import numpy as np
data1 = np.random.normal(50, 10, 100)
data2 = np.random.normal(60, 10, 100)
# Test T de Student
t_stat, p_value = stats.ttest_ind(data1, data2)

# Display the results
print("T-statistic :", t_stat)
print("P-value :", p_value)
```

```python
# Exercise 5: Linear Regression
import numpy as np
from scipy.stats import linregress
import matplotlib.pyplot as plt

# Dataset
house_sizes  = [50, 70, 80, 100, 120]
house_prices = [150000, 200000, 210000, 250000, 280000]

# Linear regression
slope, intercept, r_value, p_value, std_err = linregress(house_sizes, house_prices)

print(f"Slope      : {slope:.2f}")
print(f"Intercept  : {intercept:.2f}")
print(f"R²         : {r_value**2:.4f}")

# Prediction for 90 m²
size_to_predict = 90
predicted_price = slope * size_to_predict + intercept
print(f"Predicted price (90m²): {predicted_price:,.2f}")

# Visualization
x_line = np.linspace(40, 130, 100)
y_line = slope * x_line + intercept

plt.scatter(house_sizes, house_prices, color='blue', label='Actual Data')
plt.plot(x_line, y_line, color='red', label='Regression Line')
plt.scatter(90, predicted_price, color='green', marker='*', s=200, label='Prediction 90m²')
plt.xlabel("Size (m²)")
plt.ylabel("Price")
plt.title("Linear Regression — House Prices")
plt.legend()
plt.show()
```

```python
# Exercise 6: ANOVA Test
import numpy as np
from scipy.stats import f_oneway
import matplotlib.pyplot as plt

# Dataset
fertilizer_1 = [5, 6, 7, 6, 5]
fertilizer_2 = [7, 8, 7, 9, 8]
fertilizer_3 = [4, 5, 4, 3, 4]

# ANOVA Test
f_value, p_value = f_oneway(fertilizer_1, fertilizer_2, fertilizer_3)

print(f"F-Value : {f_value:.4f}")
print(f"P-Value : {p_value:.4f}")

if p_value < 0.05:
    print("→ Significant difference between the fertilizers.")
else:
    print("→ No significant difference between the fertilizers.")

# Visualization — Boxplot
plt.boxplot([fertilizer_1, fertilizer_2, fertilizer_3],
            labels=['Fertilizer 1', 'Fertilizer 2', 'Fertilizer 3'])
plt.ylabel("Plant Growth (cm)")
plt.title("ANOVA — Effect of Fertilizers on Plant Growth")
plt.show()
```

```python
# Exercise 7 : Probability distribution

import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import binom

# Parameters
n = 10       # number of coin flips
p = 0.5      # probability of heads (fair coin)
k = 5        # target: exactly 5 heads

# --- Question 1: Probability of exactly 5 heads ---
prob_exact = binom.pmf(k, n, p)
print(f"P(exactly 5 heads)     : {prob_exact:.4f}  ({prob_exact*100:.2f}%)")

# --- Question 2: Probability of 5 or fewer heads ---
prob_cumulative = binom.cdf(k, n, p)
print(f"P(5 or fewer heads)    : {prob_cumulative:.4f}  ({prob_cumulative*100:.2f}%)")

# --- Question 3: Probability of more than 5 heads ---
prob_more = 1 - binom.cdf(k, n, p)
print(f"P(more than 5 heads)   : {prob_more:.4f}  ({prob_more*100:.2f}%)")

# --- Full distribution: all possible outcomes (0 to 10) ---
k_values = np.arange(0, n + 1)
probabilities = binom.pmf(k_values, n, p)

# --- Visualization ---
plt.bar(k_values, probabilities, color='steelblue', edgecolor='black')
plt.bar(k, binom.pmf(k, n, p), color='red', edgecolor='black', label=f'P(X={k}) highlighted')
plt.xlabel("Number of Heads")
plt.ylabel("Probability")
plt.title("Binomial Distribution — 10 Coin Flips (p=0.5)")
plt.xticks(k_values)
plt.legend()
plt.show()
```

```python
# Exercise 8
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import pearsonr, spearmanr

# Dataset
data = pd.DataFrame({
    'age':    [23, 25, 30, 35, 40],
    'income': [35000, 40000, 50000, 60000, 70000]
})

# --- Pearson Correlation ---
pearson_corr, pearson_p = pearsonr(data['age'], data['income'])
print(f"Pearson  Coefficient : {pearson_corr:.4f}  |  P-value: {pearson_p:.4f}")

# --- Spearman Correlation ---
spearman_corr, spearman_p = spearmanr(data['age'], data['income'])
print(f"Spearman Coefficient : {spearman_corr:.4f}  |  P-value: {spearman_p:.4f}")

# --- Interpretation ---
for name, corr, p in [("Pearson", pearson_corr, pearson_p),
                       ("Spearman", spearman_corr, spearman_p)]:
    strength = "strong" if abs(corr) > 0.7 else "moderate" if abs(corr) > 0.4 else "weak"
    direction = "positive" if corr > 0 else "negative"
    significance = "significant" if p < 0.05 else "not significant"
    print(f"→ {name}: {strength} {direction} correlation, {significance} (p={p:.4f})")

# --- Visualization ---
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# Scatter plot with regression line
axes[0].scatter(data['age'], data['income'], color='steelblue', s=100, zorder=5)
m, b = np.polyfit(data['age'], data['income'], 1)
x_line = np.linspace(20, 45, 100)
axes[0].plot(x_line, m * x_line + b, color='red', linestyle='--', label='Trend line')
for i, row in data.iterrows():
    axes[0].annotate(f"({row['age']}, {row['income']:,})",
                     (row['age'], row['income']),
                     textcoords="offset points", xytext=(5, 5), fontsize=8)
axes[0].set_xlabel("Age")
axes[0].set_ylabel("Income")
axes[0].set_title(f"Scatter Plot\nPearson r = {pearson_corr:.4f}")
axes[0].legend()

# Correlation heatmap
corr_matrix = data.corr()
im = axes[1].imshow(corr_matrix, cmap='coolwarm', vmin=-1, vmax=1)
axes[1].set_xticks([0, 1])
axes[1].set_yticks([0, 1])
axes[1].set_xticklabels(['age', 'income'])
axes[1].set_yticklabels(['age', 'income'])
for i in range(2):
    for j in range(2):
        axes[1].text(j, i, f"{corr_matrix.iloc[i, j]:.4f}",
                     ha='center', va='center', fontsize=14, fontweight='bold')
plt.colorbar(im, ax=axes[1])
axes[1].set_title("Correlation Heatmap")

plt.tight_layout()
plt.show()
```
