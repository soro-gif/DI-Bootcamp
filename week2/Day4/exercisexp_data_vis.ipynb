## Exercise 1 — Why is data visualization important?

Data visualization transforms raw numbers into graphical representations that are easier to understand. It allows us to:
- **Identify trends and patterns** that would be hard to spot in a table
- **Compare values** across categories or time periods at a glance
- **Communicate results** clearly to both technical and non-technical audiences
- **Detect anomalies** (outliers) quickly

## Exercise 2 — What is the purpose of a line graph?

A **line graph** represents the evolution of a variable over an ordered sequence (usually time). It is ideal for showing **continuous trends** — rises, falls, and fluctuations — such as daily temperatures, monthly revenue, or stock prices. The connected points emphasize the continuity between values.

## Import Required Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import plotly.express as px
import plotly.graph_objects as go
```

## Example: Creating Visualizations

```python
# Create sample data for demonstration
dates = pd.date_range('2024-01-01', periods=30)
data = {
    'Date': dates,
    'Sales': np.random.randint(100, 500, 30),
    'Category': np.random.choice(['A', 'B', 'C'], 30)
}
df = pd.DataFrame(data)
print(df.head())
```

```python
# Create a line chart to show trends over time
fig = px.line(df, x='Date', y='Sales', 
               title='Sales Trend Over Time',
               markers=True)
fig.show()
```

```python
# Create a bar chart to compare categories
category_sales = df.groupby('Category')['Sales'].sum()
fig = px.bar(category_sales, 
              title='Total Sales by Category',
              labels={'index': 'Category', 'value': 'Total Sales'})
fig.show()
```
