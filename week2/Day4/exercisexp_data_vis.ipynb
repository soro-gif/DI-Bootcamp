## Exercise 1 – Understanding Data Visualization

### Why is data visualization important?

Data visualization transforms raw numbers into graphical representations that are easier to understand. It allows us to:
- **Identify trends and patterns** that would be hard to spot in a table
- **Compare values** across categories or time periods at a glance
- **Communicate results** clearly to both technical and non-technical audiences
- **Detect anomalies** (outliers) quickly

### What is the purpose of a line graph?

A **line graph** represents the evolution of a variable over an ordered sequence (usually time). It is ideal for showing **continuous trends** – rises, falls, and fluctuations – such as daily temperatures, monthly revenue, or stock prices. The connected points emphasize the continuity between values.
