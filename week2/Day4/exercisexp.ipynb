Exercise 1: Understanding Data Visualization

Task 1: Why is data visualization important in data analysis?

IMPORTANCE OF DATA VISUALIZATION

1. PATTERN RECOGNITION: Visualizations make patterns and trends visible at a glance.
   Example: A line graph immediately shows if sales are trending up or down.

2. EASIER UNDERSTANDING: Complex data becomes comprehensible through visual representation.
   Example: A pie chart shows proportions better than raw percentage numbers.

3. DECISION MAKING: Visual insights support faster and better business decisions.
   Example: Color-coded heatmaps reveal which regions need attention.

4. DATA EXPLORATION: Visualizations help discover relationships and outliers.
   Example: Scatter plots reveal correlations between variables.

5. COMMUNICATION: Visualizations communicate findings effectively to stakeholders.
   Example: A clear bar chart is more persuasive than raw data tables.

6. OUTLIER DETECTION: Visual inspection quickly identifies anomalies in data.
   Example: Box plots show outliers that statistical tables might miss.

Exercise 2: Creating a Line Plot for Temperature Variation

Objective: Create a simple line plot using Matplotlib that represents temperature variations over a week.

Implementation:

```python
import matplotlib.pyplot as plt
import numpy as np

# Temperature data for each day of the week
days = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']
temperatures = [72, 74, 76, 80, 82, 78, 75]

# Create a figure and axis
fig, ax = plt.subplots(figsize=(10, 6))

# Create the line plot
ax.plot(days, temperatures, marker='o', linewidth=2, markersize=8, color='#FF6B6B', label='Temperature')

# Label the axes
ax.set_xlabel('Day', fontsize=12, fontweight='bold')
ax.set_ylabel('Temperature (°F)', fontsize=12, fontweight='bold')

# Add a title
ax.set_title('Temperature Variation Over a Week', fontsize=14, fontweight='bold', pad=20)

# Add grid for better readability
ax.grid(True, alpha=0.3, linestyle='--')

# Add legend
ax.legend(fontsize=10)

# Rotate x-axis labels for better readability
plt.xticks(rotation=45)

# Adjust layout to prevent label cutoff
plt.tight_layout()

# Display the plot
plt.show()

print("✓ Temperature line plot created successfully!")
print(f"  - Highest temperature: {max(temperatures)}°F ({days[temperatures.index(max(temperatures))]})")
print(f"  - Lowest temperature: {min(temperatures)}°F ({days[temperatures.index(min(temperatures))]})")
print(f"  - Average temperature: {sum(temperatures)/len(temperatures):.1f}°F")
```

Exercise 3: Visualizing Monthly Sales with a Bar Chart

Objective: Generate a bar chart using Matplotlib to visualize monthly sales data for a retail store.

Implementation:

```python
import matplotlib.pyplot as plt
import numpy as np

# Monthly sales data (in dollars)
months = ['January', 'February', 'March', 'April', 'May']
sales = [5000, 5500, 6200, 7000, 7500]

# Create a figure and axis
fig, ax = plt.subplots(figsize=(10, 6))

# Create the bar chart with a gradient of colors
colors = plt.cm.viridis(np.linspace(0.3, 0.9, len(months)))
bars = ax.bar(months, sales, color=colors, edgecolor='black', linewidth=1.5, alpha=0.8)

# Label the axes
ax.set_xlabel('Month', fontsize=12, fontweight='bold')
ax.set_ylabel('Sales Amount ($)', fontsize=12, fontweight='bold')

# Add a title
ax.set_title('Monthly Sales for Retail Store', fontsize=14, fontweight='bold', pad=20)

# Add value labels on top of each bar
for i, (bar, sale) in enumerate(zip(bars, sales)):
    height = bar.get_height()
    ax.text(bar.get_x() + bar.get_width()/2., height,
            f'${sale:,}',
            ha='center', va='bottom', fontsize=10, fontweight='bold')

# Add grid for better readability (only on y-axis)
ax.grid(True, alpha=0.3, linestyle='--', axis='y')

# Set y-axis to start from 0
ax.set_ylim(0, max(sales) * 1.1)

# Rotate x-axis labels for better readability
plt.xticks(rotation=45)

# Adjust layout to prevent label cutoff
plt.tight_layout()

# Display the plot
plt.show()

# Print summary statistics
print("✓ Monthly sales bar chart created successfully!")
print(f"  - Total sales: ${sum(sales):,}")
print(f"  - Average monthly sales: ${sum(sales)/len(sales):,.2f}")
print(f"  - Highest sales: ${max(sales):,} ({months[sales.index(max(sales))]})")
print(f"  - Lowest sales: ${min(sales):,} ({months[sales.index(min(sales))]})")
print(f"  - Growth from January to May: ${sales[-1] - sales[0]:,} ({((sales[-1] - sales[0]) / sales[0] * 100):.1f}%)")
```

Exercise 4: Visualizing the Distribution of CGPA

Objective: Create a histogram to visualize the distribution of students' CGPA.

Dataset Overview: Assume the CGPA data is categorized into ranges and loaded in a DataFrame named `df`.

Tasks:
- Import necessary libraries.
- Use Seaborn's `histplot` to create a histogram of the CGPA categories.
- Customize the histogram with a specific color and add a title.
- Display the plot.

Implementation:


```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

# Create sample CGPA data (example dataset)
# In practice, this would be loaded from your actual dataset
np.random.seed(42)
cgpa_data = np.random.normal(loc=3.2, scale=0.4, size=300)
cgpa_data = np.clip(cgpa_data, 1.0, 4.0)  # CGPA typically ranges from 1.0 to 4.0

# Create a DataFrame
df = pd.DataFrame({'CGPA': cgpa_data})

# Create a figure and axis
fig, ax = plt.subplots(figsize=(10, 6))

# Create the histogram using Seaborn
sns.histplot(data=df, x='CGPA', bins=20, color='#3498db', edgecolor='black', kde=True, ax=ax)

# Customize the plot
ax.set_xlabel('CGPA', fontsize=12, fontweight='bold')
ax.set_ylabel('Frequency (Number of Students)', fontsize=12, fontweight='bold')
ax.set_title('Distribution of Students\' CGPA', fontsize=14, fontweight='bold', pad=20)

# Add grid for better readability
ax.grid(True, alpha=0.3, linestyle='--', axis='y')

# Adjust layout to prevent label cutoff
plt.tight_layout()

# Display the plot
plt.show()

# Print summary statistics
print("✓ CGPA distribution histogram created successfully!")
print(f"  - Total students: {len(df)}")
print(f"  - Mean CGPA: {df['CGPA'].mean():.2f}")
print(f"  - Median CGPA: {df['CGPA'].median():.2f}")
print(f"  - Std Dev: {df['CGPA'].std():.2f}")
print(f"  - Min CGPA: {df['CGPA'].min():.2f}")
print(f"  - Max CGPA: {df['CGPA'].max():.2f}")
```

Exercise 5: Comparing Anxiety Levels Across Different Genders

Objective: Use a bar plot to compare the proportion of students experiencing anxiety across different genders.

Dataset Overview: The dataset includes columns: 'Do you have Anxiety?' and 'Choose your gender'.

Tasks:
- Import necessary libraries.
- Use Seaborn to create a bar plot comparing anxiety levels across genders from the dataset `df`.
- Customize the plot with an appropriate color palette and add a title.
- Display the plot.

Implementation:


```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

# Create sample data for anxiety levels across genders
np.random.seed(42)
data = {
    'Choose your gender': ['Male'] * 150 + ['Female'] * 150 + ['Other'] * 100,
    'Do you have Anxiety?': (
        np.random.choice(['Yes', 'No'], 150, p=[0.35, 0.65]).tolist() +
        np.random.choice(['Yes', 'No'], 150, p=[0.48, 0.52]).tolist() +
        np.random.choice(['Yes', 'No'], 100, p=[0.42, 0.58]).tolist()
    )
}

df = pd.DataFrame(data)

# Create a figure and axis
fig, ax = plt.subplots(figsize=(10, 6))

# Create the bar plot using Seaborn
sns.countplot(data=df, x='Choose your gender', hue='Do you have Anxiety?', 
              palette='Set2', ax=ax, edgecolor='black', linewidth=1.5)

# Customize the plot
ax.set_xlabel('Gender', fontsize=12, fontweight='bold')
ax.set_ylabel('Count (Number of Students)', fontsize=12, fontweight='bold')
ax.set_title('Anxiety Levels Across Different Genders', fontsize=14, fontweight='bold', pad=20)

# Customize the legend
ax.legend(title='Do you have Anxiety?', fontsize=10, title_fontsize=11)

# Add grid for better readability
ax.grid(True, alpha=0.3, linestyle='--', axis='y')

# Adjust layout to prevent label cutoff
plt.tight_layout()

# Display the plot
plt.show()

# Print summary statistics
print("✓ Anxiety levels by gender bar plot created successfully!")
print(f"\nSummary Statistics:")
anxiety_by_gender = pd.crosstab(df['Choose your gender'], df['Do you have Anxiety?'], margins=True)
print(anxiety_by_gender)

# Calculate proportions
print(f"\nProportion with Anxiety (%):")
for gender in df['Choose your gender'].unique():
    gender_data = df[df['Choose your gender'] == gender]
    anxiety_pct = (gender_data['Do you have Anxiety?'] == 'Yes').sum() / len(gender_data) * 100
    print(f"  - {gender}: {anxiety_pct:.1f}%")
```

Exercise 6: Exploring the Relationship Between Age and Panic Attacks

Objective: Create a scatter plot to explore the relationship between students' age and the occurrence of panic attacks.

Dataset Overview: The dataset includes columns: 'Age' and 'Do you have Panic Attacks?'.

Tasks:
- Import necessary libraries.
- Convert panic attack responses to numeric values (e.g., Yes=1, No=0).
- Use Seaborn's `scatterplot` to create a scatter plot with 'Age' on the x-axis and numeric panic attack responses on the y-axis.
- Customize the plot to improve readability by adding labels, a title, and adjusting point styles.
- Display the plot.

Implementation:


```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

# Create sample data for age and panic attacks
np.random.seed(42)
ages = np.random.uniform(18, 65, 300)
panic_attacks_response = np.random.choice(['Yes', 'No'], 300, p=[0.35, 0.65])

# Convert panic attack responses to numeric values (Yes=1, No=0)
panic_attacks_numeric = [1 if response == 'Yes' else 0 for response in panic_attacks_response]

# Create a DataFrame
df = pd.DataFrame({
    'Age': ages,
    'Do you have Panic Attacks?': panic_attacks_response,
    'Panic_Attacks_Numeric': panic_attacks_numeric
})

# Create a figure and axis
fig, ax = plt.subplots(figsize=(10, 6))

# Create the scatter plot using Seaborn
sns.scatterplot(data=df, x='Age', y='Panic_Attacks_Numeric', hue='Do you have Panic Attacks?',
                palette=['#2ecc71', '#e74c3c'], size=100, alpha=0.6, ax=ax, edgecolor='black', linewidth=0.5)

# Customize the plot
ax.set_xlabel('Age (years)', fontsize=12, fontweight='bold')
ax.set_ylabel('Panic Attacks (1 = Yes, 0 = No)', fontsize=12, fontweight='bold')
ax.set_title('Relationship Between Age and Panic Attacks', fontsize=14, fontweight='bold', pad=20)

# Set y-axis ticks to show 0 and 1
ax.set_yticks([0, 1])
ax.set_yticklabels(['No', 'Yes'])

# Customize the legend
ax.legend(title='Panic Attacks', fontsize=10, title_fontsize=11, loc='upper left')

# Add grid for better readability
ax.grid(True, alpha=0.3, linestyle='--')

# Add some jitter to the y-axis points to better visualize overlapping data
for i in range(len(df)):
    df.loc[i, 'Panic_Attacks_Numeric'] += np.random.normal(0, 0.02)

# Redraw the plot with jittered data
ax.clear()
sns.scatterplot(data=df, x='Age', y='Panic_Attacks_Numeric', hue='Do you have Panic Attacks?',
                palette=['#2ecc71', '#e74c3c'], size=100, alpha=0.6, ax=ax, edgecolor='black', linewidth=0.5)

ax.set_xlabel('Age (years)', fontsize=12, fontweight='bold')
ax.set_ylabel('Panic Attacks (1 = Yes, 0 = No)', fontsize=12, fontweight='bold')
ax.set_title('Relationship Between Age and Panic Attacks', fontsize=14, fontweight='bold', pad=20)
ax.set_yticks([0, 1])
ax.set_yticklabels(['No', 'Yes'])
ax.legend(title='Panic Attacks', fontsize=10, title_fontsize=11, loc='upper left')
ax.grid(True, alpha=0.3, linestyle='--')

# Adjust layout to prevent label cutoff
plt.tight_layout()

# Display the plot
plt.show()

# Print summary statistics
print("✓ Age vs Panic Attacks scatter plot created successfully!")
print(f"\nSummary Statistics:")
print(f"  - Total students: {len(df)}")
print(f"  - Age range: {df['Age'].min():.1f} - {df['Age'].max():.1f} years")
print(f"  - Mean age: {df['Age'].mean():.1f} years")

panic_count = (df['Do you have Panic Attacks?'] == 'Yes').sum()
no_panic_count = (df['Do you have Panic Attacks?'] == 'No').sum()
print(f"\nPanic Attacks Distribution:")
print(f"  - Yes: {panic_count} ({panic_count/len(df)*100:.1f}%)")
print(f"  - No: {no_panic_count} ({no_panic_count/len(df)*100:.1f}%)")

# Calculate mean age for each group
yes_panic_age = df[df['Do you have Panic Attacks?'] == 'Yes']['Age'].mean()
no_panic_age = df[df['Do you have Panic Attacks?'] == 'No']['Age'].mean()
print(f" Mean Age by Panic Attacks Status:")
print(f"  - With Panic Attacks: {yes_panic_age:.1f} years")
print(f"  - Without Panic Attacks: {no_panic_age:.1f} years")
```
