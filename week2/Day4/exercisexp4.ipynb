Exercise 1 — Understanding Data Visualization

Why is data visualization important?

Data visualization transforms raw numbers into graphical representations that are easier to understand. It allows us to:
- **Identify trends and patterns** that would be hard to spot in a table
- **Compare values** across categories or time periods at a glance
- **Communicate results** clearly to both technical and non-technical audiences
- **Detect anomalies** (outliers) quickly

What is the purpose of a line graph?

A **line graph** represents the evolution of a variable over an ordered sequence (usually time). It is ideal for showing **continuous trends** — rises, falls, and fluctuations — such as daily temperatures, monthly revenue, or stock prices. The connected points emphasize the continuity between values.

Exercise 2 — Line Plot: Temperature Variation Over a Week

```python
import matplotlib.pyplot as plt

# Data
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
temperatures = [72, 74, 76, 80, 82, 78, 75]

# Plot
fig, ax = plt.subplots(figsize=(8, 5))
ax.plot(days, temperatures, marker='o', color='tomato', linewidth=2.5, markersize=8)

# Labels & title
ax.set_xlabel("Day", fontsize=12)
ax.set_ylabel("Temperature (°F)", fontsize=12)
ax.set_title("Temperature Variation Over a Week", fontsize=14, fontweight='bold')
ax.grid(axis='y', linestyle='--', alpha=0.6)

plt.tight_layout()
plt.show()
```

Exercise 3 — Bar Chart: Monthly Sales

```python
import matplotlib.pyplot as plt

# Data
months = ["January", "February", "March", "April", "May"]
sales = [5000, 5500, 6200, 7000, 7500]

# Plot
fig, ax = plt.subplots(figsize=(8, 5))
bars = ax.bar(months, sales, color='steelblue', edgecolor='white', width=0.6)
ax.bar_label(bars, fmt='$%d', padding=4, fontsize=10)

# Labels & title
ax.set_xlabel("Month", fontsize=12)
ax.set_ylabel("Sales Amount ($)", fontsize=12)
ax.set_title("Monthly Sales for a Retail Store", fontsize=14, fontweight='bold')
ax.set_ylim(0, 8500)
ax.grid(axis='y', linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()
```

```python
import pandas as pd

# Load dataset (place the CSV in the same folder as this notebook)
df = pd.read_csv("Student Mental health.csv")
df.columns = df.columns.str.strip()
print(df.shape)
df.head()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

cgpa_order = sorted(df['What is your CGPA?'].dropna().unique())

fig, ax = plt.subplots(figsize=(9, 5))
sns.countplot(
    data=df,
    x='What is your CGPA?',
    color='mediumpurple',
    order=cgpa_order,
    ax=ax
)

ax.set_xlabel("CGPA Range", fontsize=12)
ax.set_ylabel("Number of Students", fontsize=12)
ax.set_title("Distribution of Students' CGPA", fontsize=14, fontweight='bold')
plt.xticks(rotation=30, ha='right')
plt.tight_layout()
plt.show()
```

```python
# Exercice 5

import seaborn as sns
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(8, 5))
sns.countplot(
    data=df,
    x='Choose your gender',
    hue='Do you have Anxiety?',
    palette={'Yes': '#e76f51', 'No': '#2a9d8f'},
    edgecolor='white',
    ax=ax
)

ax.set_xlabel("Gender", fontsize=12)
ax.set_ylabel("Number of Students", fontsize=12)
ax.set_title("Anxiety Levels Across Different Genders", fontsize=14, fontweight='bold')
ax.legend(title="Has Anxiety?", fontsize=10)
ax.grid(axis='y', linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()
```

```python
# Exercice 6

import seaborn as sns
import matplotlib.pyplot as plt

# Convert Yes/No to numeric
df['Panic_Numeric'] = df['Do you have Panic attack?'].map({'Yes': 1, 'No': 0})

fig, ax = plt.subplots(figsize=(9, 5))
sns.scatterplot(
    data=df,
    x='Age',
    y='Panic_Numeric',
    hue='Do you have Panic attack?',
    palette={'Yes': '#e63946', 'No': '#457b9d'},
    style='Do you have Panic attack?',
    markers={'Yes': 'X', 'No': 'o'},
    s=80,
    alpha=0.7,
    ax=ax
)

ax.set_yticks([0, 1])
ax.set_yticklabels(['No (0)', 'Yes (1)'], fontsize=11)
ax.set_xlabel("Age", fontsize=12)
ax.set_ylabel("Panic Attack (0 = No, 1 = Yes)", fontsize=12)
ax.set_title("Relationship Between Age and Panic Attacks", fontsize=14, fontweight='bold')
ax.legend(title="Panic Attack?", fontsize=10)
ax.grid(linestyle='--', alpha=0.4)

plt.tight_layout()
plt.show()
```
