````python
#Exercice 2
from pathlib import Path
import zipfile

import pandas as pd

DATA_FOLDERS = [
    Path.cwd(),
    Path.cwd() / "DailyChalenge",
    Path.cwd() / "week2" / "Day1" / "DailyChalenge",
    Path("week2") / "Day1" / "DailyChalenge",
]

def dataset_path(filename):
    for folder in DATA_FOLDERS:
        path = folder / filename
        if path.exists():
            return path
    raise FileNotFoundError(f"Impossible de trouver {filename}")

def read_zipped_csv(zip_filename, csv_name=None):
    zip_path = dataset_path(zip_filename)
    with zipfile.ZipFile(zip_path) as archive:
        csv_files = [name for name in archive.namelist() if name.lower().endswith('.csv')]
        if csv_name is None:
            if len(csv_files) != 1:
                raise ValueError(f"Choisis un fichier CSV dans {zip_filename}: {csv_files}")
            csv_name = csv_files[0]

        with archive.open(csv_name) as csv_file:
            return pd.read_csv(csv_file, low_memory=False)

datasets = {
    "How Much Sleep Do Americans Really Get?": read_zipped_csv("How Much Sleep Do Americans.zip"),
    "Global Trends in Mental Health Disorder": read_zipped_csv("Global Trends in Mental Health Disorder.zip"),
    "Credit Card Approvals (Clean Data)": read_zipped_csv(
        "Credit Card Approvals (Clean Data).zip",
        "Credit Card Approvals (Clean Data)/clean_dataset.csv"
    ),
}

sleep_df = datasets["How Much Sleep Do Americans Really Get?"]
mental_health_df = datasets["Global Trends in Mental Health Disorder"]
credit_card_df = datasets["Credit Card Approvals (Clean Data)"]

for name, df in datasets.items():
    print(f"\n--- {name} ---")
    print("First 5 rows:")
    print(df.head())
    print("\nInfo:")
    df.info()
````

# Exercise 5: Basic Observation Skills in Data Analysis

For this exercise, I used the **How Much Sleep Do Americans Really Get?** dataset. The goal is to identify columns that could be useful for a specific type of analysis.

| Column | Interesting for | Explanation |
|---|---|---|
| Year | Trend analysis | This column is useful because it lets us observe changes in sleep habits over time. For example, we can check whether Americans sleep more or less as the years pass. |
| Avg hrs per day sleeping | Main measurement / trend analysis | This is the most important quantitative column because it contains the average number of hours slept per day. It can be used to calculate averages, compare groups, and observe trends. |
| Age Group | Group comparison | This column is useful for comparing sleep habits between different age groups. For example, we can compare teenagers, adults, and older adults. |
| Sex | Group comparison | This column allows comparison between male, female, and both groups. It can help identify whether sleep duration differs by sex. |
| Type of Days | Group comparison | This column can show whether people sleep differently depending on the type of day, such as weekdays, weekends, holidays, or all days. |
| Standard Error | Reliability check | This column helps evaluate how reliable or precise the average sleep value is. A smaller standard error usually means the estimate is more precise. |

## Possible analysis ideas

A good trend analysis would use **Year** and **Avg hrs per day sleeping** to see how average sleep changed over time. A good group comparison would use **Age Group**, **Sex**, or **Type of Days** together with **Avg hrs per day sleeping** to compare different groups.

In conclusion, the most important column for analysis is **Avg hrs per day sleeping**, because it is the main measurement. The columns **Year**, **Age Group**, **Sex**, and **Type of Days** are useful because they explain how sleep differs across time and across groups.


# Exercise 6: Identifying Data Types

| Data source | Type | Reasoning |
|---|---|---|
| A company's financial reports stored in an Excel file | Structured data | The data is organized into rows, columns, tables, and cells. This makes it easy to filter, sort, calculate totals, and analyze with tools like Excel or Pandas. |
| Photographs uploaded to a social media platform | Unstructured data | Images do not naturally fit into rows and columns. They need image processing or computer vision techniques to extract useful information. |
| A collection of news articles on a website | Unstructured data | Articles are mainly text written in paragraphs. The content does not follow a fixed table structure, even if the website itself may store some metadata. |
| Inventory data in a relational database | Structured data | Relational databases store data in organized tables with predefined columns, data types, and relationships. |
| Recorded interviews from a market research study | Unstructured data | Audio recordings are not organized in rows and columns. They must usually be transcribed or processed before analysis. |

## Summary

Structured data is organized in a clear format, usually tables with rows and columns. Unstructured data does not have a fixed tabular format and includes formats such as images, articles, and audio recordings.

# Daily Challenge: Real-world Data Analysis Scenarios

## Chosen case

The case I chose is **Heritage Grocers Group using data analysis and AI-powered analytics to improve decision-making after rapid business expansion**. Heritage Grocers Group operates grocery banners such as Cardenas Markets, El Rancho Supermercado, Tony's Fresh Market, and Los Altos Ranch Market. Between 2022 and 2024, the company expanded through acquisitions and needed a better way to understand customers across different regions and store brands.

## Context of the story

After acquiring new stores, Heritage Grocers Group had data spread across different enterprise resource planning systems and store banners. Each banner had its own reports, definitions, and customer patterns. This made it difficult to quickly answer important business questions, such as which products customers preferred in Texas compared with Chicago, or how newly acquired stores were performing.

To solve this problem, Heritage Grocers Group worked with Adastra and Microsoft. They used Microsoft Fabric, Azure OpenAI Service, and Adastra's Lucid Data Hub to unify, organize, and analyze point-of-sale data. According to Microsoft, the solution processed **1.3 terabytes of POS data** and around **3 billion item transactions**.

## What data was analyzed?

The main data analyzed was point-of-sale transaction data. This included information about products sold, purchase patterns, store locations, historical sales, customer behavior, and data from different ERP systems. Because the company had multiple store banners and regions, the data needed to be cleaned, standardized, and combined into one reliable source.

## Methods used

The project used several data analysis methods and technologies:

- Data integration: combining POS data from different systems and store banners.
- Data cleaning and standardization: making sure the same business concepts had consistent definitions across the company.
- Automated classification: organizing years of historical transaction data more quickly than manual work.
- Data warehousing and analytics: using Microsoft Fabric to create a unified data platform.
- AI-powered insights: using Azure OpenAI Service and Lucid Data Hub to accelerate analysis and support decision-making.
- Customer segmentation: analyzing purchasing behavior by region, store banner, and customer group.

## Outcomes and decisions driven by the analysis

The analysis helped Heritage Grocers Group understand customer preferences faster and more accurately. This was especially important because different communities had different shopping habits. With better data, the company could make stronger decisions about inventory, product selection, store operations, and customer service.

The project also had measurable business results. Microsoft reported that Heritage Grocers Group increased data acquisition speed, integrated data from acquired stores, reached 115 stores, processed 1.3 terabytes of POS data, and saved at least **$500,000**. The company also gained a single source of truth for operational and sales reporting.

## What would have been different without data analysis?

Without data analysis, Heritage Grocers Group would likely have relied on slower manual reporting and disconnected spreadsheets. This would have delayed decisions after acquisitions and made it harder to compare stores, regions, and customer segments. The company might have missed local purchasing trends, stocked products less effectively, or taken longer to integrate acquired stores.

Data analysis changed the situation by turning scattered transaction records into useful business insight. Instead of simply collecting sales data, the company used it to understand customers, support growth, and make faster decisions.

## Significance of data analysis

This case shows that data analysis is not only about creating charts or reports. In a real business context, data analysis can directly influence strategy, operations, customer satisfaction, and cost savings. For Heritage Grocers Group, the value came from transforming raw POS data into insights that managers could use to make decisions during a period of rapid expansion.

The most important lesson is that data analysis helps organizations move from guessing to evidence-based decision-making. By analyzing transaction data at scale, Heritage Grocers Group could better understand what customers wanted and respond more quickly to business changes.
