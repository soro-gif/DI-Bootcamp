# Titanic Dataset Preprocessing Exercises

This notebook follows one consistent preprocessing narrative:

1. Load the Titanic dataset.
2. Remove duplicate rows.
3. Handle missing values.
4. Engineer useful features.
5. Detect and treat outliers.
6. Encode categorical variables.
7. Create age groups.
8. Scale numerical features after outlier treatment.

Place `train.csv` in this folder or anywhere inside the `DI_Bootcamp` workspace before running the notebook.

```python
from pathlib import Path
import zipfile
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import MinMaxScaler, StandardScaler

pd.set_option("display.max_columns", None)
sns.set_theme(style="whitegrid")
```

## Load The Dataset

The loader searches from the current notebook folder upward. It can load either a normal `train.csv` file or the `train.csv` stored inside the local Titanic zip file.

```python
project_root = Path.cwd()
for parent in [Path.cwd(), *Path.cwd().parents]:
    if parent.name == "DI_Bootcamp":
        project_root = parent
        break

csv_matches = list(project_root.rglob("train.csv"))

if csv_matches:
    csv_path = csv_matches[0]
    raw_df = pd.read_csv(csv_path)
    loaded_from = str(csv_path)
else:
    zip_matches = list(project_root.rglob("*.zip"))
    train_member = None
    zip_path = None

    for candidate_zip in zip_matches:
        with zipfile.ZipFile(candidate_zip) as zf:
            train_files = [name for name in zf.namelist() if name.endswith("train.csv")]
            if train_files:
                zip_path = candidate_zip
                train_member = train_files[0]
                break

    if train_member is None:
        raise FileNotFoundError("Could not find train.csv or a zip file containing train.csv inside DI_Bootcamp.")

    with zipfile.ZipFile(zip_path) as zf:
        with zf.open(train_member) as file:
            raw_df = pd.read_csv(file)
    loaded_from = f"{zip_path} -> {train_member}"

df = raw_df.copy()

print(f"Loaded: {loaded_from}")
print(f"Shape: {df.shape}")
df.head()
```

## Exercise 1: Duplicate Detection And Removal

We check duplicates across all columns because the objective is to remove fully repeated records, not passengers who happen to share a name or ticket class.

```python
rows_before = len(df)
duplicate_mask = df.duplicated()
duplicate_count = duplicate_mask.sum()

print(f"Rows before duplicate removal: {rows_before}")
print(f"Duplicate rows found: {duplicate_count}")

df = df.drop_duplicates().reset_index(drop=True)
rows_after = len(df)

print(f"Rows after duplicate removal: {rows_after}")
print(f"Rows removed: {rows_before - rows_after}")
```

## Exercise 2: Handling Missing Values

Strategy used in the main preprocessing dataframe:

- `Age`: impute with the median age within each `Sex` and `Pclass` group because age patterns differ by passenger class and gender.
- `Embarked`: fill with the mode because it is a low-missingness categorical feature.
- `Cabin`: keep the missingness signal by creating `HasCabin`, then fill missing cabin labels with `Unknown`.

The examples also show `dropna()`, `fillna()`, and `SimpleImputer`, but we avoid dropping many rows from the modeling dataframe because Titanic is a small dataset.

```python
missing_summary = (
    df.isna().sum()
    .to_frame("missing_count")
    .assign(missing_percent=lambda x: (x["missing_count"] / len(df) * 100).round(2))
    .query("missing_count > 0")
    .sort_values("missing_count", ascending=False)
)

missing_summary
```

```python
# Demonstration only: dropping every row with any missing value can remove too much data.
dropna_demo = df.dropna()
print(f"Rows if using dropna(): {len(dropna_demo)} of {len(df)}")

# Demonstration only: SimpleImputer can be useful for numeric pipelines.
age_imputer = SimpleImputer(strategy="median")
age_imputed_demo = age_imputer.fit_transform(df[["Age"]])
print(f"Overall median age from SimpleImputer: {age_imputer.statistics_[0]:.1f}")
```

```python
# Main missing-value treatment.
df["Age"] = df["Age"].fillna(df.groupby(["Sex", "Pclass"])["Age"].transform("median"))
df["Age"] = df["Age"].fillna(df["Age"].median())

embarked_mode = df["Embarked"].mode(dropna=True)[0]
df["Embarked"] = df["Embarked"].fillna(embarked_mode)

df["HasCabin"] = df["Cabin"].notna().astype(int)
df["Cabin"] = df["Cabin"].fillna("Unknown")

df.isna().sum().sort_values(ascending=False).head(10)
```

## Exercise 3: Feature Engineering

We create compact features that are common in Titanic modeling:

- `FamilySize`: passenger plus siblings/spouses and parents/children.
- `IsAlone`: binary flag derived from family size.
- `Title`: extracted from the passenger name and grouped to reduce rare categories.
- `Deck`: first cabin letter, with `U` for unknown cabins.

```python
df["FamilySize"] = df["SibSp"] + df["Parch"] + 1
df["IsAlone"] = (df["FamilySize"] == 1).astype(int)

df["Title"] = df["Name"].str.extract(r",\s*([^\.]+)\.", expand=False).str.strip()

# Group rare or socially equivalent titles to avoid creating many sparse dummy columns.
title_map = {
    "Mlle": "Miss",
    "Ms": "Miss",
    "Mme": "Mrs",
    "Lady": "Rare",
    "Countess": "Rare",
    "Capt": "Rare",
    "Col": "Rare",
    "Don": "Rare",
    "Dr": "Rare",
    "Major": "Rare",
    "Rev": "Rare",
    "Sir": "Rare",
    "Jonkheer": "Rare",
    "Dona": "Rare",
}
df["Title"] = df["Title"].replace(title_map)

df["Deck"] = df["Cabin"].str[0].replace("U", "Unknown")

df[["Name", "SibSp", "Parch", "FamilySize", "IsAlone", "Title", "Cabin", "Deck"]].head()
```

## Exercise 4: Outlier Detection And Handling

`Fare` is strongly right-skewed, so we compare IQR outliers and high quantiles, then use 98th-percentile capping plus `log1p`. The 98th percentile keeps almost all observations while limiting the influence of the most extreme fares.

`Age` is naturally bounded and less extreme, so we inspect it with the same method and only cap if the IQR rule finds unrealistic high values.

```python
fig, axes = plt.subplots(2, 2, figsize=(12, 8))

sns.histplot(df["Fare"], kde=True, ax=axes[0, 0])
axes[0, 0].set_title("Fare Before Treatment")
sns.boxplot(x=df["Fare"], ax=axes[0, 1])
axes[0, 1].set_title("Fare Boxplot")

sns.histplot(df["Age"], kde=True, ax=axes[1, 0])
axes[1, 0].set_title("Age Before Treatment")
sns.boxplot(x=df["Age"], ax=axes[1, 1])
axes[1, 1].set_title("Age Boxplot")

plt.tight_layout()
```

```python
def iqr_bounds(series, multiplier=1.5):
    """Return standard Tukey IQR bounds for outlier detection."""
    q1 = series.quantile(0.25)
    q3 = series.quantile(0.75)
    iqr = q3 - q1
    return q1 - multiplier * iqr, q3 + multiplier * iqr

for column in ["Fare", "Age"]:
    lower, upper = iqr_bounds(df[column])
    outlier_count = ((df[column] < lower) | (df[column] > upper)).sum()
    print(f"{column}: IQR bounds=({lower:.2f}, {upper:.2f}), outliers={outlier_count}")

df[["Fare", "Age"]].quantile([0.95, 0.98, 0.99, 1.00])
```

```python
df_before_outliers = df.copy()

fare_cap = df["Fare"].quantile(0.98)
age_lower, age_upper = iqr_bounds(df["Age"])

# Fare has rare extreme values, so cap at the empirical 98th percentile and log-transform.
df["Fare_Capped"] = df["Fare"].clip(upper=fare_cap)
df["Fare_Log"] = np.log1p(df["Fare_Capped"])

# Age is bounded by human life span; clipping to IQR bounds is conservative after median imputation.
df["Age_Capped"] = df["Age"].clip(lower=max(0, age_lower), upper=age_upper)

comparison = pd.DataFrame({
    "before_fare": df_before_outliers["Fare"].describe(),
    "after_fare_capped": df["Fare_Capped"].describe(),
    "before_age": df_before_outliers["Age"].describe(),
    "after_age_capped": df["Age_Capped"].describe(),
})

comparison
```

## Exercise 6: Feature Encoding

We encode nominal categorical variables with one-hot encoding. There are no true ordinal categorical variables here, so label encoding is not needed for the final dataset.

```python
categorical_cols = ["Sex", "Embarked", "Title", "Deck"]

encoded_df = pd.get_dummies(
    df,
    columns=categorical_cols,
    drop_first=True,
    dtype=int,
)

encoded_columns = [col for col in encoded_df.columns if any(col.startswith(f"{base}_") for base in categorical_cols)]
print(f"Created {len(encoded_columns)} encoded columns")
encoded_df[encoded_columns].head()
```

## Exercise 7: Data Transformation For Age Feature

We create life-stage bins with `pd.cut()` and one-hot encode them. The bins are simple and interpretable: child, teen, adult, and senior.

```python
age_bins = [0, 12, 18, 60, 100]
age_labels = ["Child", "Teen", "Adult", "Senior"]

encoded_df["AgeGroup"] = pd.cut(
    encoded_df["Age_Capped"],
    bins=age_bins,
    labels=age_labels,
    include_lowest=True,
)

age_group_dummies = pd.get_dummies(encoded_df["AgeGroup"], prefix="AgeGroup", dtype=int)
encoded_df = pd.concat([encoded_df.drop(columns=["AgeGroup"]), age_group_dummies], axis=1)

age_group_dummies.head()
```

## Exercise 5: Data Standardization And Normalization

Scaling is done after outlier treatment. We standardize features that are roughly continuous and model-distance sensitive, and we min-max scale skewed or bounded features.

- `Age_Capped`: standardized because age is less skewed after imputation and capping.
- `Fare_Log`: standardized after log transformation because the log greatly reduces skew.
- `FamilySize`: min-max scaled because it is bounded and count-based.

```python
final_df = encoded_df.copy()

standard_cols = ["Age_Capped", "Fare_Log"]
minmax_cols = ["FamilySize", "SibSp", "Parch"]

standard_scaler = StandardScaler()
minmax_scaler = MinMaxScaler()

final_df[[f"{col}_Std" for col in standard_cols]] = standard_scaler.fit_transform(final_df[standard_cols])
final_df[[f"{col}_MinMax" for col in minmax_cols]] = minmax_scaler.fit_transform(final_df[minmax_cols])

final_df[[
    "Age_Capped_Std",
    "Fare_Log_Std",
    "FamilySize_MinMax",
    "SibSp_MinMax",
    "Parch_MinMax",
]].describe().round(3)
```

## Final Modeling Dataset

For a modeling table, we drop high-cardinality text identifiers and original columns that have already been transformed. The target `Survived` is kept.

```python
columns_to_drop = [
    "PassengerId",
    "Name",
    "Ticket",
    "Cabin",
    "Age",
    "Fare",
    "Fare_Capped",
    "Age_Capped",
    "Fare_Log",
    "SibSp",
    "Parch",
    "FamilySize",
]

model_df = final_df.drop(columns=[col for col in columns_to_drop if col in final_df.columns])

print(f"Final modeling shape: {model_df.shape}")
model_df.head()
```
